```python
from scipy import stats
import numpy as np

continuous_cols = ['thalach', 'oldpeak']

# Calculate Z-scores
z_scores = np.abs(stats.zscore(df[continuous_cols]))

# Filter out rows where Z-score is greater than 3
df = df[(z_scores < 3).all(axis=1)]
```

```python
df.describe()
```

```python
df.corr()['target']
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_train
x_test
y_train
y_test
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(x_train,y_train)
```

```python
y_pred=clf.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
age = int(input("Age: "))
sex = int(input("Sex (0=female, 1=male): "))
cp = int(input("Chest pain type (0-3): "))
trestbps = int(input("Resting blood pressure: "))
chol = int(input("Cholesterol: "))
fbs = int(input("Fasting blood sugar > 120 (0/1): "))
restecg = int(input("Resting ECG (0-2): "))
thalach = int(input("Max heart rate: "))
exang = int(input("Exercise induced angina (0/1): "))
oldpeak = float(input("ST depression: "))
slope = int(input("Slope (0-2): "))
ca = int(input("Number of major vessels (0-4): "))
thal = int(input("Thal (1-3): "))

new_patient = pd.DataFrame({
    'age':[age],'sex':[sex],'cp':[cp],'trestbps':[trestbps],'chol':[chol],
    'fbs':[fbs],'restecg':[restecg],'thalach':[thalach],'exang':[exang],
    'oldpeak':[oldpeak],'slope':[slope],'ca':[ca],'thal':[thal]
})

new_patient_scaled = scaler.transform(new_patient)
pred = clf.predict(new_patient_scaled)
print("Result:", "Heart Disease" if pred[0]==1 else "No Heart Disease")
```
